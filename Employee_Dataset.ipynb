```python
df_cleaned = df.dropna()
df_cleaned
```

```python
df["salary"] = df['salary'].fillna(df['salary'].median())
df["salary"]
```

```python
print("Duplicate Rows:", df.duplicated().sum())
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.rename(
    columns={
        'healthy_eating': 'healthy_eating_score',
        'active_lifestyle': 'active_lifestyle_score',
    },
    inplace=True
)

df.rename
```

```python
print(df['groups'].unique())
```

```python
print(df['groups'].value_counts())
```

```python
df['groups'] = df['groups'].str.strip().str.upper()
df['groups'] = df['groups'].replace({
    'GROUP A': 'GROUP_A',
    'GROUP B': 'GROUP_B',
    'GROUP C': 'GROUP_C',
    'GROUP D': 'GROUP_D',
    'GROUP E': 'GROUP_E',
    'GROUP F': 'GROUP_F',
    'GROUP G': 'GROUP_G',
    'GROUP H': 'GROUP_H',
    'GROUP I': 'GROUP_I',
    'GROUP J': 'GROUP_J',
    'GROUP K': 'GROUP_K',
    'GROUP L': 'GROUP_L',
    'GROUP M': 'GROUP_M',
    'GROUP N': 'GROUP_N',
    'GROUP O': 'GROUP_O'
})
df['groups']
```

```python
df['age'] = df['age'].astype(int)
df['id'] = df['id'].astype(int)
df['salary'] = df['salary'].astype(float)
df['age']
df['id']
df['salary']
```

```python
q1 = df['salary'].quantile(0.25)
q3 = df['salary'].quantile(0.75)
iqr = q3 - q1

lower_bound = q1 - 1.5 * iqr
upper_bound = q3 + 1.5 * iqr

df['salary'] = df['salary'].clip(lower=lower_bound, upper=upper_bound)
df['salary']
df['salary']
```

```python
median_salary = df['salary'].median()
filtered_df = df[(df['age'] > 30) & (df['salary'] > median_salary)]
filtered_df
```

```python
df['lifestyle_score'] = (
    df['healthy_eating_score'] + df['active_lifestyle_score']
)
df['lifestyle_score']
```

```python
df_sorted = df.sort_values(by='salary', ascending=False)
df_sorted
```

```python
df.to_csv('cleaned_employee_dataset.csv', index=False)
print("Data cleaning completed and exported successfully!")
```

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv('Employee Dataset.csv')
```

```python
plt.figure(figsize=(10,6))
plt.hist(df['salary'], bins=30, color='skyblue', edgecolor='black', alpha=0.9)
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.title('Distribution of Employee Salaries', fontsize=16, fontweight='bold')
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(
    df['age'], df['salary'], color='purple', alpha=0.7, edgecolors='k', s=70
)

plt.title('Employee Age vs. Salary', fontsize=14, fontweight='bold')
plt.xlabel('Age', fontsize=12)
plt.ylabel('Salary', fontsize=12)
plt.grid(True, linestyle='--', alpha=0.5)

plt.show()
```

```python
avg_salary_by_group = df.groupby('groups')['salary'].mean()

plt.figure(figsize=(12,6))
avg_salary_by_group.plot(
    kind='bar',color='teal',edgecolor ='black', alpha=0.8
)

plt.title('Average Salary by Group', fontsize=16, fontweight='bold')
plt.xlabel('Groups', fontsize=12)
plt.ylabel('Average Salary', fontsize=12)
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(6, 5))
plt.boxplot(
    df['salary'],
    patch_artist=True,
    boxprops=dict(facecolor='orange', color='black'),
    medianprops=dict(color='red', linewidth=2),
)

plt.title('Salary Outlier Detection (Box Plot)', fontsize=14, fontweight='bold')
plt.ylabel('Salary', fontsize=12)
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.show()
```
